### Two pointers Explanation:

1. You start with two pointers—often at the start and end of the array/string.

2. In each iteration, you compare or process the elements at these pointers.

3. Then move one or both pointers based on conditions relevant to the problem.

4. Repeat until pointers meet or cross.

5. O(n) time

```python
from typing import List 


def twoSum(numbers: List[int], target: int) -> List[int]:

    left_pointer = 0
    right_pointer = len(numbers) - 1

    while left_pointer < right_pointer:
        sum_nums = numbers[left_pointer] + numbers[right_pointer]

        if sum_nums == target:
            return [left_pointer + 1,  right_pointer + 1]
        
        elif sum_nums > target:
            right_pointer -= 1
        elif sum_nums < target:
            left_pointer += 1
    
    return []

    
twoSum(numbers = [2,7,11,15], target = 9)
```

```python
def maxArea(height) -> int:
    
    left_pointer = 0
    right_pointer = len(height) - 1
    max_area = 0

    while left_pointer < right_pointer:
        width = right_pointer - left_pointer
        area = min(height[left_pointer], height[right_pointer]) * width
        max_area = max(max_area, area)

        if height[left_pointer] < height[right_pointer]:
            left_pointer += 1
        else:
            right_pointer -= 1 
            
    return max_area
maxArea(height = [1,8,6,2,5,4,8,3,7])
```

```python
def threeSum(array):

    array.sort()

    result = []

    for i in range(len(array)):
        if array[i] == array[i-1] and i > 0:
            continue

        left_pointer = i + 1
        right_pointer = len(array) - 1
         
        while left_pointer < right_pointer: 
            sums = array[i] + array[left_pointer] + array[right_pointer]

            if sums == 0:
                result.append([array[i], array[left_pointer], array[right_pointer]])
            
                left_pointer += 1
                right_pointer -= 1
                
                while left_pointer < right_pointer and array[left_pointer] == array[left_pointer - 1]:
                    left_pointer += 1
                while left_pointer < right_pointer and array[right_pointer] == array[right_pointer + 1]:
                    right_pointer -= 1

            elif sums < 0:
                left_pointer += 1
            else:
                right_pointer -= 1
    return result 

threeSum(array=[-1,0,1,2,-1,-4])
```

```python
def fourSum(nums, target):

    nums.sort()
    
    result = []

    for i in range(len(nums)):
        if i > 0 and nums[i] == nums[i - 1]:
            continue 

        for j in range(i + 1, len(nums)):
            if j > i + 1 and nums[j] == nums[j-1]:
                continue 

            left_pointer = j + 1
            right_pointer = len(nums) - 1

            while left_pointer < right_pointer:
                sums = nums[i] + nums[j] + nums[left_pointer] + nums[right_pointer]

                if sums == target:
                    result.append((nums[i], nums[j], nums[left_pointer], nums[right_pointer]))

                    left_pointer += 1
                    right_pointer -= 1

                    # remove duplicates
                
                    while left_pointer < right_pointer and nums[left_pointer] == nums[left_pointer - 1]:
                        left_pointer += 1
                    
                    while left_pointer < right_pointer and nums[right_pointer] == nums[right_pointer + 1]:
                        right_pointer -= 1
                
                elif sums < target:
                    left_pointer += 1
                else:
                    right_pointer -= 1
    return result


fourSum(nums=[1,0,-1,0,-2,2], target=0)
```

```python
def rolling_7_day_window(data):

    if len(data) < 7:
        return []
    
    window = 0
    left = 0 
    result = []

    for right in range(len(data)):
        window += data[right]
        print(window)

        if right - left + 1 == 7:
            result.append(window)
            window -= data[left]
            left += 1

    return result 
print(rolling_7_day_window(data=[20,10,30,18,40,41,12]))
```

```python
def rolling_14_day_window(data, target):

    if len(data) < target:
        return [] 
    
    window = 0
    left = 0 
    result = [] 

    for right in range(len(data)):
        window += data[right] # incrementally add all numbers till the end of len(data)

        if right - left + 1 == target: # when it reaches the target window append results
            result.append(window)
            window -= data[left] # always remove the first element of the window
            left += 1
    return result 

print(rolling_14_day_window([20,10,30,18,40,41,12,20,10,30,18,40,41,12], target = 14))
```

```python
def findMaxAverage(nums, k) -> float:

    if len(nums) < k:
        return 0.0
    
    left = 0 
    window = 0
    result = float('inf')

    for right in range(len(nums)):
        window += nums[right]


        if right - left + 1 == k:
            result = min(result, window)
            window -= nums[left]
            left += 1

            return result/k 
        else:
            return nums[0]

print(findMaxAverage(nums = [20,10,30,18,40,41,12,20,10,30,18,40,41,12], k = 14))
```

```python
#leetcode 977

def sortedSquares(nums):

    left = 0
    right = len(nums) - 1
    
    result = []
    while left <= right: 
        if abs(nums[left]) > abs(nums[right]): # use abs for negative nums (-4)^2
            result.append(nums[left]**2)
            left += 1
        else:
            result.append(nums[right]**2)
            right -= 1
    
    return result[::-1]

sortedSquares(nums=[-4,-1,0,3,10])
```
